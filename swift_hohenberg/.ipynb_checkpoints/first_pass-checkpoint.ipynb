Integrate SH35 (and maybe SH23) forward in time with Dirichlet BC versus periodic BC to observe selected patterns.

```python
import numpy as np
import matplotlib.pyplot as plt
import animation
import sys
np.random.seed(0)
```

```python
def construct_lap_periodic(nx):
    '''Construct laplacian operator with periodic BCs'''
    
    lap = -2*np.diag(np.ones(nx)) + 1*np.diag(np.ones(nx-1), 1) + 1*np.diag(np.ones(nx-1), -1)
    
    lap[0,-1] = 1    # Periodic BCs
    lap[-1,0] = 1 
    
    return lap

def construct_lap_dirichlet(nx):
    '''Construct laplacian operator with dirichlet BCs (zeros in corners)'''
    
    lap = -2*np.diag(np.ones(nx)) + 1*np.diag(np.ones(nx-1), 1) + 1*np.diag(np.ones(nx-1), -1)

    
    return lap
```

```python
# iteration params
dt = 0.00001
nt = 10000

dx = .10625 # near the boundary
nx = 100
x = np.linspace(0, dx*nx, nx)

# PDE params
r = 100.0
qc = 3

b3 = 0.1 
b5 = 0.1

params = (r, qc, b3, b5)
u0 = np.random.rand(nx)
# u0 = np.cos(x*qc)

# integrate forward in time
def integrate_explicit(u0, params, nx, dx, nt, dt, bc='dirichlet'):
    '''Integrate SH35 according to explicit Euler using specified BCs.'''
    
    r, qc, b3, b3 = params
    if bc=='dirichlet':
        lap_op = construct_lap_dirichlet(nx) / dx**2
    elif bc=='periodic':
        lap_op = construct_lap_periodic(nx) / dx**2
    qc_op = np.eye(nx) * qc
    
    d = 0
    lap_op_2 = lap_op * d # new diffusive operator
    
    # initialize field
    u_stored = np.zeros((nx, nt))
    u_stored[:,0] = u0
    
    # step forward in time
    for i in range(1, nt):
        
        if int(100*i/nt) > int(100*(i-1)/nt):
            sys.stdout.write("\rIteration {0}/{1} ({2}%)".format(i, nt, int(100*i/nt)))
            sys.stdout.flush()
            
        u = u_stored[:,i-1]
        
        nonlin = b3*u**3 - b5*u**5
        harmonic_op = np.linalg.matrix_power(qc_op@qc_op + lap_op, 2) #+ lap_op_2
        
        u_stored[:,i] = u + dt*(r*u - harmonic_op@u + nonlin)
        if bc=='dirichlet': u_stored[0,i] = 0; u_stored[-1,i] = 0  
    
        # if i==1:
        #     print(r*u)
        #     print(-harmonic_op@u)
        
    return u_stored
```

```python
u_stored = integrate_explicit(u0, params, nx, dx, nt, dt)
```

```python
# plt.plot(u_stored[:,3])
# plt.show()
animation.animate_2d_array(u_stored[:,::400])
```

Then look at steady-state fourier distribution to see how selected wavenumber changes with boundary conditions.

```python
uf = np.fft.fft(u_stored[:,-1])
uf2 = abs(uf)**2
# plt.plot(uf2)

ks = np.fft.fftfreq(nx, dx)[np.argmax(uf2)]*2*np.pi # selected wavenumber in rad/m

ks
```

Loop over several dx values and analyze selected wavelength

```python
ndx = 10
dx_array = np.linspace(.1, .3, ndx)

# iteration params
dt = 0.00001
nt = 30000

nx = 100
x = np.linspace(0, dx*nx, nx)

# PDE params
r = 100.0
qc = 3

b3 = 0.1 
b5 = 0.1

params = (r, qc, b3, b5)
u0 = np.random.rand(nx)

# Open and erase contents of data file
with open("data.txt", "w") as f:
    f.write("")
    
    
for idx in range(ndx):
    if idx>0: u0 = u_stored[:,-1] # start with previous steady-state
    print(idx)
    
    dx = dx_array[idx]
    u_stored = integrate_explicit(u0, params, nx, dx, nt, dt, bc='dirichlet')
    
    with open("data.txt", "a") as f:
        f.write("params = {}, nx = {}, dx = {}, nt = {}, dt = {} \n".format(params, nx, dx, nt, dt))
        np.savetxt(f, u_stored[:,-1], newline=',')
        f.write("\n")
    
#     plt.plot(u_stored[:,-1], )
    
# plt.show()
```

Read stored data to get selected wavenumber vs L

```python
# open and read data
datafile = 'data_.1to.3_adiabatic.txt'
ndx = 100

# only take even lines to read u data
data = [];
cnt = 1;
with open(datafile, 'r') as infile:
    for line in infile:
        if cnt%2 == 0:
            data.append(line.split(','))
        cnt+=1
data = np.asarray(np.array(data)[:,:-1], dtype=float)
    

# only take odd lines to read params data
pdata = [];
cnt = 1;
with open(datafile, 'r') as infile:
    for line in infile:
        if cnt%2 == 1:
            pdata.append(line.split(','))
        cnt+=1
# extract dx data
dxdata = np.zeros(ndx)
for idx in range(ndx):
    dxdata[idx] = pdata[idx][5][5:]      
        

# fourier transform u data
fdata = np.fft.fft(data, axis=1)
fdata2 = abs(fdata)**2

# find selected wavenumbers
ks = np.zeros(ndx)
for idx in range(ndx):
    nx = 100 # be sure this is true, shouldn't be hardcoded
    dx = dxdata[idx]
    ks[idx] = np.fft.fftfreq(nx, dx)[np.argmax(fdata2[idx])]*2*np.pi # selected wavenumber in rad/m

    
```
